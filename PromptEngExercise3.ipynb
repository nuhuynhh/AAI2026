```python
import os
from getpass import getpass
import google.generativeai as genai

os.environ["GEMINI_API_KEY"] = getpass("Enter your Gemini API key: ")
genai.configure(api_key=os.environ["GEMINI_API_KEY"])

model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
source_text = """
AI is increasingly used in customer support, helping companies respond faster and more consistently.
However, automated support can frustrate users when it provides generic answers, misunderstands the issue,
or fails to escalate complex cases. Effective AI support systems often combine automation with clear
handoff rules to human agents, maintain a professional tone, and ask targeted clarifying questions.
Organizations must also monitor these systems for bias, privacy risks, and hallucinated information.
A well-designed support AI should prioritize accuracy, transparency, and user trust while still improving efficiency.
""".strip()
print(source_text)
```

```python
before_prompt = f"""
Summarize the following text in 120–150 words. Write in paragraph form.

Text:
{source_text}
""".strip()

before_resp = model.generate_content(before_prompt)
before_summary = before_resp.text.strip()

print("BEFORE SUMMARY:\n")
print(before_summary)
```

```python
critique_prompt = f"""
You are going to self-critique and revise your own summary.

Original text:
{source_text}

Your current summary:
{before_summary}

Critique the summary using these criteria:
1) Accuracy: Does it match the source without adding new claims?
2) Clarity: Is it easy to understand for a college student?
3) Completeness: Does it include the main benefits AND risks?
4) Conciseness: Keep the revision between 90–110 words.
5) Format: Use 3 bullet points (no paragraph). Each bullet must be one sentence.

First, write a brief critique (3–5 bullets).
Then provide a revised summary that satisfies ALL criteria.
""".strip()

critique_resp = model.generate_content(critique_prompt)
critique_and_after = critique_resp.text.strip()

print(critique_and_after)
```

```python
after_only_prompt = f"""
From the critique-and-revise process, output ONLY the revised summary.
Requirements:
- 3 bullet points
- 90–110 words total
- No extra commentary
""".strip()

after_resp = model.generate_content(after_only_prompt)
after_summary = after_resp.text.strip()

print("AFTER SUMMARY:\n")
print(after_summary)
```
